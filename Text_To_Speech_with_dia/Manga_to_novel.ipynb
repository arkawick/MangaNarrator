```python
import os
import easyocr
from PIL import Image
from google.colab import files
import google.generativeai as genai
from gtts import gTTS
from IPython.display import Audio
```

```python
# Paste your Gemini API key here
# Get one from: https://aistudio.google.com/app/apikey
GOOGLE_API_KEY = "your_google_api_key_here"
genai.configure(api_key=GOOGLE_API_KEY)
```

```python
uploaded = files.upload()
image_path = next(iter(uploaded))
```

```python
reader = easyocr.Reader(['en'])
results = reader.readtext(image_path, detail=0)
extracted_text = "\n".join(results)

print("🗨️ Extracted Dialogue:\n", extracted_text)
```

```python
model = genai.GenerativeModel("gemini-1.5-flash-latest")

prompt = f"""Convert the following manga panel dialogue into a novel-style story, describing the scene and emotions with rich detail.

Dialogue:
{extracted_text}
"""

response = model.generate_content(prompt)
story = response.text

print("\n📖 Novel-style Story:\n")
print(story)


```

```python
tts = gTTS(text=story)
tts.save("manga_novel_story.mp3")
```
